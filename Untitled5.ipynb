```python
import os, re, time, random, shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import tensorflow as tf
from sklearn.model_selection import GroupShuffleSplit
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, confusion_matrix, classification_report, roc_curve)
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, SimpleRNN, LSTM, GRU, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

SEED = 42
random.seed(SEED); np.random.seed(SEED); tf.keras.utils.set_random_seed(SEED)
os.makedirs("figures", exist_ok=True)

def save_text_image(text, path, fs=9):
    # saves text output as a clean PNG "screenshot" for the report
    lines = text.rstrip("\n").split("\n")
    w = max(len(l) for l in lines); pad = 0.25
    wi = max(w * fs * 0.0083 + 2 * pad, 4); hi = len(lines) * fs * 0.0195 + 2 * pad
    fig = plt.figure(figsize=(wi, hi), dpi=170); fig.patch.set_facecolor("#f7f7f7")
    fig.text(pad / wi, 1 - pad / hi, "\n".join(lines), family="DejaVu Sans Mono", fontsize=fs,
             va="top", color="#202124", linespacing=1.35)
    fig.savefig(path, facecolor="#f7f7f7"); plt.close(fig)

IMG = 64                      # images are resized to IMG x IMG  ->  IMG time steps x IMG features
CLASSES = ["NORMAL", "PNEUMONIA"]   # label 0 = NORMAL, label 1 = PNEUMONIA
print("TensorFlow:", tf.__version__)
print("GPU available:", bool(tf.config.list_physical_devices("GPU")))
```

```python
# Set DATA_ROOT manually only if you uploaded the dataset yourself; otherwise leave None to use KaggleHub.
DATA_ROOT = None

def find_root(base):
    # locate the folder that contains both 'train' and 'test' (ignores __MACOSX)
    for dp, dn, _ in os.walk(base):
        if "__MACOSX" in dp: continue
        if "train" in dn and "test" in dn:
            return dp
    raise FileNotFoundError("Could not find train/test folders under " + base)

if DATA_ROOT is None:
    import kagglehub
    path = kagglehub.dataset_download("paultimothymooney/chest-xray-pneumonia")
    DATA_ROOT = find_root(path)
print("Dataset root:", DATA_ROOT)
print("Sub-folders :", sorted(os.listdir(DATA_ROOT)))
```

```python
IMG_EXT = (".jpg", ".jpeg", ".png")
rows = []
for split in ["train", "val", "test"]:
    for cls in CLASSES:
        d = os.path.join(DATA_ROOT, split, cls)
        n = len([f for f in os.listdir(d) if f.lower().endswith(IMG_EXT)]) if os.path.isdir(d) else 0
        rows.append({"split": split, "class": cls, "images": n})
counts = pd.DataFrame(rows)
pivot = counts.pivot(index="split", columns="class", values="images").assign(total=lambda t: t.sum(axis=1))
display(pivot)
save_text_image(pivot.to_string(), "figures/snap_folder_counts.png", fs=11)

plt.figure(figsize=(6, 4))
sns.barplot(data=counts, x="split", y="images", hue="class")
plt.title("Chest X-ray images per split and class"); plt.tight_layout()
plt.savefig("figures/fig1_class_distribution.png", dpi=150); plt.show()
```

```python
def patient_id(fname, cls):
    m = re.match(r"(person\d+)", fname) if cls == "PNEUMONIA" else re.search(r"(IM-\d+)", fname)
    return f"{cls}_{m.group(1)}" if m else f"{cls}_{fname}"

def load_folder(split):
    X, y, groups, bad = [], [], [], 0
    for label, cls in enumerate(CLASSES):
        d = os.path.join(DATA_ROOT, split, cls)
        if not os.path.isdir(d): continue
        for f in sorted(os.listdir(d)):
            if not f.lower().endswith(IMG_EXT): continue
            try:
                im = Image.open(os.path.join(d, f)).convert("L").resize((IMG, IMG), Image.Resampling.LANCZOS)
                X.append(np.asarray(im, dtype="float32") / 255.0)
                y.append(label); groups.append(patient_id(f, cls))
            except Exception:
                bad += 1
    print(f"{split:5s}: loaded {len(X)} images, skipped {bad} unreadable")
    return np.array(X), np.array(y), np.array(groups)

t0 = time.time()
X_tr, y_tr, g_tr = load_folder("train")
X_va0, y_va0, g_va0 = load_folder("val")
X_test, y_test, _ = load_folder("test")
print(f"Loading took {time.time()-t0:.0f}s")

# The official 'val' folder has only 16 images, so it is merged with 'train' and a new validation set is made below.
X_pool = np.concatenate([X_tr, X_va0]); y_pool = np.concatenate([y_tr, y_va0]); g_pool = np.concatenate([g_tr, g_va0])
print("Pool:", X_pool.shape, " Test:", X_test.shape)
```

```python
fig, axes = plt.subplots(2, 4, figsize=(11, 6))
for r, label in enumerate([0, 1]):
    idx = np.where(y_pool == label)[0][:4]
    for c, i in enumerate(idx):
        axes[r, c].imshow(X_pool[i], cmap="gray"); axes[r, c].axis("off")
        axes[r, c].set_title(f"{CLASSES[label]} (64 rows = 64 time steps)", fontsize=8)
plt.suptitle("Resized X-rays: each image is read row by row as a sequence")
plt.tight_layout(); plt.savefig("figures/fig2_sample_images.png", dpi=150); plt.show()
```

```python
gss = GroupShuffleSplit(n_splits=1, test_size=0.15, random_state=SEED)
tr_idx, va_idx = next(gss.split(X_pool, y_pool, groups=g_pool))
X_train, y_train = X_pool[tr_idx], y_pool[tr_idx]
X_val,   y_val   = X_pool[va_idx], y_pool[va_idx]

overlap = set(g_pool[tr_idx]) & set(g_pool[va_idx])
print("Patients shared between train and validation:", len(overlap))
split_table = pd.DataFrame({
    "set": ["Train", "Validation", "Test"],
    "NORMAL": [(y_train==0).sum(), (y_val==0).sum(), (y_test==0).sum()],
    "PNEUMONIA": [(y_train==1).sum(), (y_val==1).sum(), (y_test==1).sum()]})
split_table["total"] = split_table["NORMAL"] + split_table["PNEUMONIA"]
display(split_table)
split_table.to_csv("figures/split_table.csv", index=False)

# class weights counter the imbalance found in Week 2 (many more PNEUMONIA than NORMAL images)
w = compute_class_weight("balanced", classes=np.array([0, 1]), y=y_train)
class_weight = {0: float(w[0]), 1: float(w[1])}
print("Class weights:", class_weight)
save_text_image(f"Patients shared between train and validation: {len(overlap)}\n\n" + split_table.to_string(index=False)
                + f"\n\nClass weights: NORMAL = {class_weight[0]:.3f}, PNEUMONIA = {class_weight[1]:.3f}",
                "figures/snap_split.png", fs=11)
```

```python
def build_model(cell, units=64, dropout=0.3, lr=1e-3):
    model = Sequential([
        Input(shape=(IMG, IMG)),
        cell(units),
        Dropout(dropout),
        Dense(32, activation="relu"),
        Dense(1, activation="sigmoid"),
    ], name=cell.__name__)
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=lr, clipnorm=1.0),
                  loss="binary_crossentropy",
                  metrics=["accuracy", tf.keras.metrics.AUC(name="auc")])
    return model

summ_lines, n_params = [], {}
for cell in (SimpleRNN, LSTM, GRU):
    m = build_model(cell); n_params[cell.__name__] = m.count_params()
    lines = []
    m.summary(print_fn=lambda line, *a, **k: lines.append(line))
    summ_lines += lines + [""]
print("\n".join(summ_lines))
save_text_image("\n".join(summ_lines), "figures/snap_model_summaries.png", fs=8)
```

```python
EPOCHS, BATCH = 25, 32
models, histories, train_time = {}, {}, {}
for cell in (SimpleRNN, LSTM, GRU):
    name = cell.__name__
    tf.keras.utils.set_random_seed(SEED)
    model = build_model(cell)
    es = EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
    t0 = time.time()
    h = model.fit(X_train, y_train, validation_data=(X_val, y_val), epochs=EPOCHS, batch_size=BATCH,
                  class_weight=class_weight, callbacks=[es], verbose=2)
    train_time[name] = time.time() - t0
    models[name], histories[name] = model, h
    print(f"\n>>> {name}: {len(h.history['loss'])} epochs, best val_loss = {min(h.history['val_loss']):.4f}, time = {train_time[name]:.0f}s\n")

# screenshot of the last epochs of each model + a table with epochs / time / parameters
log = []
for name, h in histories.items():
    hh, n = h.history, len(h.history["loss"])
    log.append(f"=== {name}: {n} epochs, best val_loss = {min(hh['val_loss']):.4f}, time = {train_time[name]:.0f}s ===")
    for e in range(max(0, n - 6), n):
        log.append(f"Epoch {e+1:>2}/{EPOCHS} - loss: {hh['loss'][e]:.4f} - acc: {hh['accuracy'][e]:.4f} - auc: {hh['auc'][e]:.4f}"
                   f" - val_loss: {hh['val_loss'][e]:.4f} - val_acc: {hh['val_accuracy'][e]:.4f} - val_auc: {hh['val_auc'][e]:.4f}")
    log.append("")
save_text_image("\n".join(log), "figures/snap_training_log.png", fs=8)
train_info = pd.DataFrame({n: {"params": n_params[n], "epochs": len(histories[n].history["loss"]),
                               "best_val_loss": min(histories[n].history["val_loss"]), "train_time_s": train_time[n]}
                           for n in histories}).T
display(train_info); train_info.to_csv("figures/train_info.csv")
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 7))
for j, name in enumerate(models):
    h = histories[name].history
    axes[0, j].plot(h["loss"], label="Train"); axes[0, j].plot(h["val_loss"], label="Validation")
    axes[0, j].set_title(f"{name} – loss"); axes[0, j].set_xlabel("Epoch"); axes[0, j].legend(); axes[0, j].grid(alpha=.3)
    axes[1, j].plot(h["accuracy"], label="Train"); axes[1, j].plot(h["val_accuracy"], label="Validation")
    axes[1, j].set_title(f"{name} – accuracy"); axes[1, j].set_xlabel("Epoch"); axes[1, j].legend(); axes[1, j].grid(alpha=.3)
plt.tight_layout(); plt.savefig("figures/fig3_training_curves.png", dpi=150); plt.show()
```

```python
probs, preds, rows = {}, {}, {}
for name, model in models.items():
    p = model.predict(X_test, batch_size=64, verbose=0).ravel()
    yhat = (p >= 0.5).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, yhat, labels=[0, 1]).ravel()
    probs[name], preds[name] = p, yhat
    rows[name] = {"Accuracy": accuracy_score(y_test, yhat),
                  "Precision": precision_score(y_test, yhat, zero_division=0),
                  "Recall (sensitivity)": recall_score(y_test, yhat),
                  "Specificity": tn / (tn + fp),
                  "F1": f1_score(y_test, yhat), "ROC-AUC": roc_auc_score(y_test, p),
                  "TN": tn, "FP": fp, "FN": fn, "TP": tp,
                  "Epochs": len(histories[name].history["loss"]), "Train time (s)": train_time[name]}
results = pd.DataFrame(rows).T
for c in ["TN", "FP", "FN", "TP", "Epochs"]: results[c] = results[c].astype(int)
results = results.round(3)
display(results)
results.to_csv("figures/results_table.csv")
save_text_image(results.to_string(), "figures/snap_results_table.png", fs=9)

rep_txt = ""
for name in models:
    rep_txt += f"===== {name} =====\n" + classification_report(y_test, preds[name], target_names=CLASSES, digits=3, zero_division=0) + "\n"
print(rep_txt)
save_text_image(rep_txt, "figures/snap_classification_reports.png", fs=8)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, name in zip(axes, models):
    cm = confusion_matrix(y_test, preds[name], labels=[0, 1])
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False, ax=ax,
                xticklabels=CLASSES, yticklabels=CLASSES)
    ax.set_title(f"{name}"); ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
plt.suptitle("Confusion matrices on the test set"); plt.tight_layout()
plt.savefig("figures/fig4_confusion_matrices.png", dpi=150); plt.show()

plt.figure(figsize=(6, 5))
for name in models:
    fpr, tpr, _ = roc_curve(y_test, probs[name])
    plt.plot(fpr, tpr, label=f"{name} (AUC = {results.loc[name, 'ROC-AUC']:.3f})")
plt.plot([0, 1], [0, 1], "k--", label="Chance")
plt.xlabel("False positive rate"); plt.ylabel("True positive rate"); plt.title("ROC curves (test set)")
plt.legend(); plt.grid(alpha=.3); plt.tight_layout()
plt.savefig("figures/fig5_roc_curves.png", dpi=150); plt.show()
```
