```python
import kagglehub
import os
from pathlib import Path

dataset_path = kagglehub.dataset_download(
    "paultimothymooney/chest-xray-pneumonia"
)

print("Dataset downloaded to:")
print(dataset_path)
```

```python
for root, dirs, files in os.walk(dataset_path):
    level = root.replace(dataset_path, "").count(os.sep)

    if level < 3:
        print("  " * level + os.path.basename(root) + "/")
```

```python
import pandas as pd

data_dir = Path(dataset_path)

chest_xray = list(data_dir.rglob("chest_xray"))[0]

train_dir = chest_xray / "train"
val_dir = chest_xray / "val"
test_dir = chest_xray / "test"

records = []

for split, split_dir in [
    ("train", train_dir),
    ("val", val_dir),
    ("test", test_dir)
]:

    for class_name in ["NORMAL", "PNEUMONIA"]:

        class_dir = split_dir / class_name

        for image_path in class_dir.glob("*"):

            if image_path.suffix.lower() in [".jpeg", ".jpg", ".png"]:

                records.append({
                    "filepath": str(image_path),
                    "split": split,
                    "label": class_name
                })

df = pd.DataFrame(records)

print("Dataset shape:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nFive random samples:")
display(df.sample(5, random_state=42))
```

```python
print("Missing filepath values:")
print(df["filepath"].isna().sum())

print("\nMissing labels:")
print(df["label"].isna().sum())
```

```python
print("Duplicate file paths:")
print(df["filepath"].duplicated().sum())
```

```python
import hashlib

def file_hash(filepath):
    with open(filepath, "rb") as f:
        return hashlib.md5(f.read()).hexdigest()

df["image_hash"] = df["filepath"].apply(file_hash)

print("Total images:", len(df))
print("Unique image hashes:", df["image_hash"].nunique())
print(
    "Duplicate images:",
    len(df) - df["image_hash"].nunique()
)
```

```python
from PIL import Image

corrupt_images = []

for filepath in df["filepath"]:

    try:
        with Image.open(filepath) as img:
            img.verify()

    except Exception as e:
        corrupt_images.append((filepath, str(e)))

print(
    "Number of corrupt/unreadable images:",
    len(corrupt_images)
)
```

```python
class_counts = df["label"].value_counts()

print(class_counts)

print("\nClass percentages:")
print(
    (class_counts / len(df) * 100).round(2)
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image


def get_random_images(label, n=50):
    return df[
        df["label"] == label
    ]["filepath"].sample(
        n=n,
        random_state=42
    )


plt.figure(figsize=(10, 6))


for label in ["NORMAL", "PNEUMONIA"]:

    pixel_values = []

    paths = get_random_images(label, 50)

    for filepath in paths:

        img = Image.open(filepath).convert("L")

        img = np.array(img)

        pixel_values.extend(
            img.flatten()[::20]
        )

    plt.hist(
        pixel_values,
        bins=50,
        alpha=0.5,
        label=label
    )


plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.title(
    "Pixel Intensity Distribution by Class"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    "intensity_histogram.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
IMG_SIZE = (224, 224)

def calculate_mean_image(
    label,
    n=300
):

    paths = df[
        df["label"] == label
    ]["filepath"].sample(
        n,
        random_state=42
    )

    images = []

    for filepath in paths:

        img = Image.open(
            filepath
        ).convert("L")

        img = img.resize(
            IMG_SIZE
        )

        arr = np.array(
            img,
            dtype=np.float32
        ) / 255.0

        images.append(arr)

    return np.mean(
        images,
        axis=0
    )


mean_normal = calculate_mean_image(
    "NORMAL"
)

mean_pneumonia = calculate_mean_image(
    "PNEUMONIA"
)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.imshow(
    mean_normal,
    cmap="gray"
)

plt.title(
    "Mean Image - NORMAL"
)

plt.axis("off")

plt.subplot(1, 2, 2)

plt.imshow(
    mean_pneumonia,
    cmap="gray"
)

plt.title(
    "Mean Image - PNEUMONIA"
)

plt.axis("off")

plt.tight_layout()

plt.savefig(
    "mean_images.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import re

def extract_patient_id(filepath):

    filename = Path(filepath).stem

    match = re.match(
        r"(person\d+)",
        filename
    )

    if match:
        return match.group(1)

    return None


df["patient_id"] = df[
    "filepath"
].apply(
    extract_patient_id
)

print(
    df[
        ["filepath", "patient_id"]
    ].head()
)
```

```python
train_patients = set(
    df[
        df["split"] == "train"
    ]["patient_id"].dropna()
)

val_patients = set(
    df[
        df["split"] == "val"
    ]["patient_id"].dropna()
)

test_patients = set(
    df[
        df["split"] == "test"
    ]["patient_id"].dropna()
)

print(
    "Train patients:",
    len(train_patients)
)

print(
    "Validation patients:",
    len(val_patients)
)

print(
    "Test patients:",
    len(test_patients)
)

print(
    "\nTrain-Test overlap:"
)

print(
    train_patients.intersection(
        test_patients
    )
)

print(
    "\nTrain-Validation overlap:"
)

print(
    train_patients.intersection(
        val_patients
    )
)

print(
    "\nValidation-Test overlap:"
)

print(
    val_patients.intersection(
        test_patients
    )
)
```
